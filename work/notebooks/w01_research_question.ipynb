## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("sample_data/content_refresh_anonymized.csv")
print(f"Loaded {df.shape[0]} rows and {df.shape[1]} columns.\n")

# Filter for pages with sufficient impressions to get meaningful CTR data
visible = df[df["impressions_90d"] >= 100]

# Calculate mean CTR by position tier
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)

# Calculate key numbers
top_3_ctr = ctr_by_pos.get('top_3', 0)
page_1_ctr = ctr_by_pos.get('page_1', 0)
striking_ctr = ctr_by_pos.get('striking', 0)
page_3_5_ctr = ctr_by_pos.get('page_3_5', 0)

# Take the average of top positions for a clearer baseline
average_top_ctr = (top_3_ctr + page_1_ctr) / 2 if (top_3_ctr > 0 or page_1_ctr > 0) else 0

if average_top_ctr > 0:
    # Percentage drop from top positions to 'striking'
    drop_to_striking_pct = ((average_top_ctr - striking_ctr) / average_top_ctr) * 100
    # Percentage drop from top positions to 'page_3_5'
    drop_to_page_3_5_pct = ((average_top_ctr - page_3_5_ctr) / average_top_ctr) * 100
else:
    drop_to_striking_pct = 0
    drop_to_page_3_5_pct = 0


print(f"Mean CTR by position tier (for pages with >= 100 impressions):\n{ctr_by_pos.round(3).to_string()}\n")
print(f"**Key takeaway:** The average CTR for top positions (top_3, page_1) is around {average_top_ctr:.3f}. \
Moving from these top positions to 'striking' positions results in an average CTR drop of approximately {drop_to_striking_pct:.1f}%.\
Further, moving to 'page_3_5' positions sees an even sharper drop of about {drop_to_page_3_5_pct:.1f}%.")
print("This highlights the critical importance of achieving and maintaining top search rankings for driving clicks.")

corr_search_impressions = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between `search_volume` and `impressions_90d`: {corr_search_impressions:.3f}\n")
print("**Key takeaway:** A correlation near 0.00 suggests that a keyword's search volume \
is barely predictive of the actual impressions (traffic) a page receives. \
This indicates that simply targeting high search volume keywords isn't sufficient for success, \
underscoring the need for more nuanced, ML-driven content optimization.")

total_pages = len(df)
declining_pages = df[df['trend_direction'] == 'down']
num_declining_pages = len(declining_pages)
percentage_declining = (num_declining_pages / total_pages) * 100

print(f"Total pages in dataset: {total_pages}")
print(f"Pages with 'down' trend_direction: {num_declining_pages}")
print(f"Percentage of pages with a 'down' trend: {percentage_declining:.1f}%\n")
print(f"**Key takeaway:** Approximately {percentage_declining:.1f}% of pages are currently experiencing a declining trend.\
This large number represents a significant opportunity for improvement through content refresh and optimization strategies, \
which is precisely what the ML pipeline aims to address over the next 7 weeks.")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

# **Observed**
*   Performance Gaps & Organic Trends: Performance varies significantly across pages, with over half (53.9%) experiencing declining impressions and specific pages lagging well behind their expected position-tier click-through rate (CTR) benchmarks.
*   Metric Misalignment: Keyword search volume shows almost no correlation with real impressions, proving that volume metrics alone fail to reliably predict actual traffic potential or search visibility.

# **Directional**
*   Targeted CTR Optimization: Pages in striking distance positions or with large gaps below tier averages represent high-value opportunities where snippet improvements can yield significant incremental click growth.
*   Prioritized Action: Combining negative impression trends with below-benchmark CTRs establishes an opportunity score that isolates the highest-urgency pages requiring immediate intervention.

# **Decision-support**
*   Actionable Prioritization Framework: The data provides a dynamic, benchmark-backed system that ranks snippet-rewrite opportunities by recoverable clicks, sets target CTR benchmarks for content teams, and flags weekly score changes to maintain focus on the highest-leverage pages.

# **What this will never do**
*   Because the model only observes gap data rather than conducting controlled experiments, it cannot directly attribute post-rewrite CTR changes to specific optimizations. Subsequent performance shifts are easily influenced by external variables—such as seasonality, competitor movement, algorithmic updates, or Google's dynamic SERP rendering—that introduce noise into the results. Consequently, establishing true attribution requires a statistically rigorous before-and-after measurement window, which falls entirely outside the design scope of this model.
*   Operating strictly within observed ranking and click behavior, the model identifies performance gaps without predicting algorithm-driven ranking changes or attempting to model their underlying causes. Furthermore, its CTR benchmarks are internally derived from this specific site and time window, making them locally valid for prioritization rather than externally generalizable across different industries or websites.
*   It will never present the opportunity score as an exact prediction of recoverable clicks. The score is an ordinal ranking signal, not a revenue forecast. Saying "rewriting this page will recover 847 clicks" implies a precision the data cannot support.
